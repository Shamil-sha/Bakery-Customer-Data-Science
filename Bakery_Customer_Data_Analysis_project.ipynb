```python
# Select input features
X = df_cleaned[
    ["Customer_Age", "Customer_Gender", "Customer_Segment"]
].copy()

# Select the target column
y = df_cleaned["Loyalty_Member"].copy()

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nInput features:")
display(X.head())

print("\nTarget values:")
print(y.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Identify numerical and categorical features
numeric_features = ["Customer_Age"]

categorical_features = [
    "Customer_Gender",
    "Customer_Segment"
]

# Convert categorical values into numerical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical",
         OneHotEncoder(handle_unknown="ignore"),
         categorical_features)
    ],
    remainder="passthrough"
)

X_encoded = preprocessor.fit_transform(X)

print("Original feature shape:", X.shape)
print("Encoded feature shape:", X_encoded.shape)

print("\nEncoded feature names:")
print(preprocessor.get_feature_names_out())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "Customer_Gender",
    "Customer_Segment"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("categorical",
         OneHotEncoder(handle_unknown="ignore"),
         categorical_features)
    ],
    remainder="passthrough"
)

X_encoded = preprocessor.fit_transform(X)

print("Original feature shape:", X.shape)
print("Encoded feature shape:", X_encoded.shape)

print("\nEncoded feature names:")
print(preprocessor.get_feature_names_out())
```

```python
from sklearn.model_selection import train_test_split

# Split the data: 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create the model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
from sklearn.metrics import accuracy_score

# Predict loyalty membership for test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Detailed performance report
print("Classification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=["No", "Yes"])

print("\nConfusion Matrix:")
print(cm)

# Visualize the confusion matrix
display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No", "Yes"]
)

display.plot()
plt.title("Loyalty Membership - Confusion Matrix")
plt.tight_layout()
plt.show()
```
